```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
sp.init_printing()
x,y=sp.symbols("x y",real=True)
```

## 1. Symbolic partial derivatives
Compute first and second derivatives, then compare the mixed partials.

```python
f=x**2*sp.exp(x*y)+sp.sin(y)
fx,fy=sp.diff(f,x),sp.diff(f,y)
display(fx,fy,sp.simplify(sp.diff(fx,y)-sp.diff(fy,x)))
```

## 2. Level curves and surface graphs

```python
a=np.linspace(-2,2,120)
X,Y=np.meshgrid(a,a)
Z=2+.6*X**2+.9*Y**2
fig,ax=plt.subplots()
c=ax.contour(X,Y,Z,levels=10)
ax.clabel(c);ax.set_aspect("equal");plt.show()
```

## 3. Tangent-plane error
For $f(x,y)=x^2+2y^2$ at $(1,1/2)$, compare the exact function with its tangent plane.

```python
xx,yy=np.meshgrid(np.linspace(.5,1.5,80),np.linspace(0,1,80))
exact=xx**2+2*yy**2
linear=1.5+2*(xx-1)+2*(yy-.5)
err=exact-linear
print("Maximum error:",np.max(np.abs(err)))
plt.contourf(xx,yy,err,levels=18);plt.colorbar(label="Exact - linear");plt.show()
```

## 4. Chain rule check

```python
t=sp.symbols("t",real=True)
g=x**2*sp.exp(x*y)+sp.sin(y)
path={x:t**2,y:sp.sin(t)}
by_substitution=sp.diff(g.subs(path),t)
by_chain=sp.diff(g,x).subs(path)*2*t+sp.diff(g,y).subs(path)*sp.cos(t)
assert sp.simplify(by_substitution-by_chain)==0
print("Chain rule verified")
```

## 5. Finite-difference convergence
Compare the forward and centered approximations for $f_x(0.8,0.6)$.

```python
fn=lambda a,b:a*a+2*b*b
steps=10.**(-np.arange(1,9,dtype=float))
forward=np.array([(fn(.8+h,.6)-fn(.8,.6))/h for h in steps])
central=np.array([(fn(.8+h,.6)-fn(.8-h,.6))/(2*h) for h in steps])
plt.loglog(steps,np.abs(forward-1.6),"o-",label="Forward")
plt.loglog(steps,np.maximum(np.abs(central-1.6),1e-16),"s-",label="Centered")
plt.xlabel("Step size");plt.ylabel("Absolute error");plt.legend();plt.show()
```
