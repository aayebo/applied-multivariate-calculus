```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from scipy.optimize import minimize
x, y = sp.symbols("x y", real=True)
```

## 1. Gradient and Hessian
Compute the gradient and Hessian of $f(x,y)=e^x\cos(y)+xy$ at the origin. Explain the mixed term.

```python
f = sp.exp(x)*sp.cos(y)+x*y
grad = sp.Matrix([sp.diff(f,x),sp.diff(f,y)])
H = sp.hessian(f,(x,y))
print("Gradient:", grad)
print("Hessian:", H)
```

## 2. Taylor error
Compare the exact function to its quadratic approximation at several points. What happens when distance from the origin increases?

```python
v = sp.Matrix([x,y])
T2 = sp.expand(f.subs({x:0,y:0})+(grad.subs({x:0,y:0}).T*v)[0]+(v.T*H.subs({x:0,y:0})*v)[0]/2)
for a,b in [(0.1,0.2),(0.3,-0.2),(0.8,0.8)]:
    exact=float(f.subs({x:a,y:b})); approx=float(T2.subs({x:a,y:b}))
    print((a,b),"exact",exact,"quadratic",approx,"error",abs(exact-approx))
```

## 3. Gradient descent
Vary the step size. Which choices converge? Use the Hessian to explain why.

```python
def descent(alpha, steps=40):
    p=np.array([-2.,2.]);path=[p.copy()]
    for _ in range(steps):
        p-=alpha*np.array([2*(p[0]-2),8*(p[1]+1)])
        path.append(p.copy())
    return np.array(path)
for a in [.05,.15,.35]:
    path=descent(a)
    print(a, path[-1])
    plt.plot(path[:,0],path[:,1],"-o",markersize=2,label=f"step={a}")
plt.scatter([2],[-1],color="black",marker="*",s=90)
plt.xlabel("x");plt.ylabel("y");plt.legend();plt.show()
```

## 4. SciPy optimization and least squares
Check the numerical solution and compare it with the exact minimizer.

```python
def objective(v): return (v[0]-2)**2+4*(v[1]+1)**2
r=minimize(objective,[-2,2],method="BFGS")
print(r.x,r.success,r.message)
xs=np.array([0.,1.,2.,3.]);ys=np.array([1.,2.,2.,4.])
A=np.column_stack([np.ones_like(xs),xs])
print("Least-squares coefficients:",np.linalg.lstsq(A,ys,rcond=None)[0])
```
