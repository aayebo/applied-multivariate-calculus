# Chapter 5: Multiple Integration and Applications

Explore exact integrals, numerical quadrature, Jacobians, and Monte Carlo estimation. All visualizations and code are original.

```python
import numpy as np
import sympy as sp
from scipy.integrate import dblquad, tplquad
import matplotlib.pyplot as plt
```

## 1. Two integration orders
Compute the same integral over a triangular region using both orders.

```python
x,y=sp.symbols("x y",real=True)
f=x+y
vertical=sp.integrate(f,(y,0,x),(x,0,1))
horizontal=sp.integrate(f,(x,y,1),(y,0,1))
print(vertical,horizontal)
```

## 2. Polar volume
The paraboloid $z=4-x^2-y^2$ over the disk $x^2+y^2\le4$ has volume $8\pi$.

```python
value,err=dblquad(lambda r,t:(4-r*r)*r,0,2*np.pi,lambda t:0,lambda t:2)
print(value, 8*np.pi,err)
```

## 3. Mass and centroid
Use $\rho(x,y)=1+x$ on $[0,2]\times[0,1]$.

```python
rho=1+x
M=sp.integrate(rho,(y,0,1),(x,0,2))
xbar=sp.integrate(x*rho,(y,0,1),(x,0,2))/M
ybar=sp.integrate(y*rho,(y,0,1),(x,0,2))/M
print("mass",M,"centroid",(xbar,ybar))
```

## 4. Jacobians

```python
u,v=sp.symbols("u v",real=True)
T=sp.Matrix([u+v,u-v])
print(T.jacobian([u,v]).det())
```

## 5. Monte Carlo convergence
Repeat with increasing sample sizes and compare with the exact $8\pi$.

```python
rng=np.random.default_rng(2026)
ns=[100,1000,10000,100000]
est=[];ses=[]
for N in ns:
    xy=rng.uniform(-2,2,size=(N,2))
    r2=(xy**2).sum(axis=1)
    g=np.where(r2<=4,4-r2,0)
    est.append(16*g.mean())
    ses.append(16*g.std(ddof=1)/np.sqrt(N))
plt.errorbar(ns,est,yerr=[2*s for s in ses],fmt="o-",capsize=4,label="MC estimate ± 2 SE")
plt.axhline(8*np.pi,linestyle="--",label="Exact")
plt.xscale("log");plt.xlabel("Number of samples");plt.ylabel("Volume");plt.legend();plt.show()
```
