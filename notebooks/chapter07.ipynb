```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
sp.init_printing()
```

## 1. Compute the metric of a cylinder

```python
u,v,a=sp.symbols("u v a",real=True,positive=True)
X=sp.Matrix([a*sp.cos(u),a*sp.sin(u),v])
J=X.jacobian([u,v])
g=sp.simplify(J.T*J)
g
```

## 2. Compute polar Christoffel symbols

```python
r,th=sp.symbols("r th",positive=True)
q=(r,th);g=sp.diag(1,r**2);gi=g.inv()
Gamma=[[[sp.simplify(sum(gi[k,l]*(sp.diff(g[l,j],q[i])+sp.diff(g[l,i],q[j])-sp.diff(g[i,j],q[l]))/2 for l in range(2))) for j in range(2)] for i in range(2)] for k in range(2)]
Gamma
```

## 3. Solve a Euclidean straight line using polar geodesic equations

```python
def rhs(t,y):
    r,th,vr,vt=y
    return [vr,vt,r*vt**2,-2*vr*vt/r]
sol=solve_ivp(rhs,(0,2),[1,0,0,1],rtol=1e-10,atol=1e-12)
x=sol.y[0]*np.cos(sol.y[1]);y=sol.y[0]*np.sin(sol.y[1])
fig,ax=plt.subplots()
ax.plot(x,y,label="numerical geodesic")
ax.plot(np.ones_like(sol.t),sol.t,"--",label="exact straight line")
ax.set_aspect("equal");ax.legend();plt.show()
print("maximum error",np.max(np.abs(y-sol.t)))
```

## 4. Explore curvature of two graphs

```python
x,y=sp.symbols("x y",real=True)
for f in [x*x+y*y,x*x-y*y]:
    K=sp.simplify((sp.diff(f,x,2)*sp.diff(f,y,2)-sp.diff(f,x,y)**2)/(1+sp.diff(f,x)**2+sp.diff(f,y)**2)**2)
    print(f,"K =",K)
```
