```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from scipy.integrate import quad, dblquad
sp.init_printing()
```

## 1. Visualize circulation and radial flux

```python
x, y = np.meshgrid(np.linspace(-2, 2, 17), np.linspace(-2, 2, 17))
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].quiver(x, y, -y, x); ax[0].set_title("Rotational field")
ax[1].quiver(x, y, x, y); ax[1].set_title("Radial field")
for a in ax: a.set_aspect("equal"); a.grid(alpha=.2)
plt.show()
```

## 2. Potential function and path independence

```python
X, Y = sp.symbols("X Y", real=True)
phi = X**2*Y + 3*X + Y**2
F = sp.Matrix([sp.diff(phi,X),sp.diff(phi,Y)])
print("Field:", F.T)
print("Work (0,0) to (1,2):",phi.subs({X:1,Y:2})-phi.subs({X:0,Y:0}))
```

## 3. Green’s theorem: direct circulation versus curl integral

```python
boundary, be = quad(lambda t: np.sin(t)**2 + np.cos(t)**2, 0, 2*np.pi)
interior, ie = dblquad(lambda r, th: 2*r, 0, 2*np.pi, lambda th: 0, lambda th: 1)
print("Boundary:", boundary, "Interior:", interior, "Difference:", abs(boundary-interior))
```

## 4. Divergence theorem on a sphere

```python
a=2.0
flux_surface = a * (4*np.pi*a*a)
flux_volume = 3*(4*np.pi*a**3/3)
print(flux_surface,flux_volume)
```

## 5. Stokes’ theorem: disk and hemisphere with the same boundary

```python
disk, _ = dblquad(lambda r, th: r, 0, 2*np.pi, lambda th: 0, lambda th: 1)
hemisphere, _ = dblquad(lambda ph, th: np.sin(ph)*np.cos(ph), 0, 2*np.pi, lambda th: 0, lambda th: np.pi/2)
print("Disk flux:",disk,"Hemisphere flux:",hemisphere,"Boundary circulation:",np.pi)
```
