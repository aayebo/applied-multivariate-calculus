## Lab 2: Dot product and angle
Compare an exact symbolic expression with a numerical angle.

```python
u = sp.Matrix([1,2,2]); v = sp.Matrix([2,0,1])
cos_theta = sp.simplify(u.dot(v)/(u.norm()*v.norm()))
print("cos(theta) =", cos_theta)
print("theta (degrees) =", float(sp.acos(cos_theta)*180/sp.pi))
```

## Lab 3: A plane from three points
Compute a normal and the closest point to the plane. Explain the geometry of the perpendicular displacement.

```python
A,B,C = (sp.Matrix(q) for q in [(1,0,1),(2,1,1),(1,2,3)])
P=sp.Matrix([3,1,4])
n=(B-A).cross(C-A)/2
d=n.dot(A)
Q=sp.simplify(P-(n.dot(P)-d)/n.dot(n)*n)
print("normal:", n.T, "closest point:", Q.T)
assert n.dot(Q)==d
```

## Lab 4: Visualize the plane and its normal
Rotate the view and change the transparency. Which aspects are mathematical invariants?

```python
x,y=np.meshgrid(np.linspace(-1,3,25),np.linspace(-1,3,25))
z=2-x+y
fig=plt.figure(figsize=(7,5));ax=fig.add_subplot(111,projection="3d")
ax.plot_surface(x,y,z,alpha=.45,cmap="Blues")
ax.quiver(1,0,1,1,-1,1,color="firebrick",length=1.5,normalize=True)
ax.set(xlabel="x",ylabel="y",zlabel="z")
plt.show()
```
