```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
from scipy.integrate import quad
```

## 1. Arc length: exact versus numerical
Find the length of $r(t)=(t,t^2)$ for $0\leq t\leq1$.

```python
t = sp.symbols("t", real=True)
exact = sp.integrate(sp.sqrt(1 + 4*t**2), (t,0,1))
numeric, estimated_error = quad(lambda u: np.sqrt(1+4*u*u),0,1)
print("Exact:",exact,"Numerical:",numeric,"Difference:",float(exact)-numeric)
```

## 2. Curvature and torsion of a helix
Derive and interpret the two invariants.

```python
a,b=sp.symbols("a b",positive=True)
r=sp.Matrix([a*sp.cos(t),a*sp.sin(t),b*t])
v=r.diff(t); acc=r.diff(t,2); jerk=r.diff(t,3)
c=v.cross(acc)
kappa=sp.simplify(sp.sqrt(sp.simplify(c.dot(c)))/sp.sqrt(sp.simplify(v.dot(v)))**3)
tau=sp.simplify(c.dot(jerk)/c.dot(c))
print("curvature",kappa,"torsion",tau)
```

Change the coefficient of the cosine and describe how the shape changes.

## 4. Parametrized surface and tangent plane
Verify that the surface normal is perpendicular to both tangent directions.

```python
u,v=sp.symbols("u v",real=True)
S=sp.Matrix([u,v,u*u-v*v]); Su=S.diff(u);Sv=S.diff(v);normal=Su.cross(Sv)
print("normal:",normal,"dot products:",sp.simplify(normal.dot(Su)),sp.simplify(normal.dot(Sv)))
```

## 5. Numerical convergence investigation
Estimate the arc length of a helix by polygons. Graph error against the number of subintervals on logarithmic axes.

```python
curve=lambda t:np.array([2*np.cos(t),2*np.sin(t),.5*t])
ns=np.array([20,40,80,160,320,640])
exact=4*np.pi*np.sqrt(4.25)
errors=[]
for n in ns:
    xyz=np.array([curve(s) for s in np.linspace(0,4*np.pi,n+1)])
    errors.append(exact-np.linalg.norm(np.diff(xyz,axis=0),axis=1).sum())
plt.loglog(ns,errors,"o-");plt.xlabel("Number of segments");plt.ylabel("Absolute error");plt.grid(True,which="both");plt.show()
```
